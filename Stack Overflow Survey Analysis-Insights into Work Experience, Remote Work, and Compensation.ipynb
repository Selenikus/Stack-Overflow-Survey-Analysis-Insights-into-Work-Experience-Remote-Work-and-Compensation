# 2. Soru:

#### Kaç kişi tüm soruları yanıtladı?


- Survey_results_schema.csv, anketin yapısını açıklarken survey_results_public.csv, gerçek anket yanıtlarını içeriyor.
- Survey_results_schema.csv içindeki qname sütunu ile survey_results_public.csv içindeki sütun isimleri birebir eşleşiyor.
- Survey_results_schema.csv dosyasında 87 satır var, demek ki ankette sorulan gerçek soru sayısı 87.
- Survey_results_public.csv dosyasında 114 sütun var, yani fazladan 114 - 87 = 27 sistem bilgisi içeren sütun var.
- Tüm soruları yanıtlayanları bulmak için, sadece 87 soru sütununa odaklanıp eksik veri olup olmadığını kontrol ettim.

```python
## Schema Veri Seti'nden geçerli qname satırlarını aldım.

qname_columns = df['qname'].dropna()
qname_columns
```

```python
##  Public'de mevcut olan ve Shema ile eşleşen sütunları seçtim.

valid_columns = [x for x in qname_columns if x in df2.columns]

# NaN değerlerini, geçerli anket sorularının olduğu sütunlarda kontrol edip, satırları getirmesini istedim.

complete_count = df2.dropna(subset=valid_columns).shape[0]
print(f"Tüm soruları yanıtlayan katılımcı sayısı: {complete_count}")
```

# 5. Soru

#### Katılımcıların yüzde kaçı Python ile programlama yapıyor?

* Public veri setindeki sütunları 'df2.columns.tolist()' kodu tararayarak 'LanguageHaveWorkedWith' kolonunu buldum.

```python
## - Verisetinin ilgili sütunun içinde string bir ifade olan 'Python'ı aratıp, na: olan değerlesin false olarak sayılmasını sağlayan,
# True gelen tüm satırları sayan bir kod yazdım.Toplam 'Python' kullanıcı sayısı 30795.

python_users = python_users = df2[df2['LanguageHaveWorkedWith'].str.contains('Python', na=False)].shape[0]
python_users
```

```python
# Python kullanıcılarını, 1. soruda hesapladığım tüm katılımcılara bölerek yüzdesini aldım.

percentage_python = (python_users / katilimci_sayisi1) * 100
print(f"Katılımcıların yüzde {percentage_python:.2f}'si Python ile programlama yapıyor.")
```

#### Ülkelere göre gruplanmış ve Python ile programlama yapan katılımcılar arasında, her ülkede ortalama ve medyan yıllık tazminat (ConvertedCompYearly) ne kadardır?
